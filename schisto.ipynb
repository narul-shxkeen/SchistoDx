```python
model = YOLO("yolo11l.pt")  # load a pretrained model (recommended for training)
```

```python
results = model.train(data="./Schistosomiases-Eggs--6/data.yaml", epochs=200, imgsz=640, batch=2)
```

```python
# Cell for validation with confusion matrix
result = YOLO("./runs/detect/train/weights/best.pt")  # load the best trained model weights

# Validate the model
metrics = result.val()  # validate using the loaded model

# Print metrics
print(f"mAP50-95: {metrics.box.map}")
print(f"mAP50: {metrics.box.map50}")
print(f"mAP75: {metrics.box.map75}")

# Print confusion matrix
print("\nConfusion Matrix:")
print(metrics.confusion_matrix.matrix)

# Optionally, plot and save the confusion matrix
metrics.confusion_matrix.plot(save_dir='./runs/detect/train', names=list(metrics.names.values()))
```
